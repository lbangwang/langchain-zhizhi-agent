1.提示词模版PromptTemplate和ChatPromptTemplate

```python
import base64
from langchain.chat_models import init_chat_model
from dotenv import load_dotenv
import os

from langchain_core.prompts import PromptTemplate, ChatPromptTemplate

load_dotenv(override=True)

DASHSCOPE_API_KEY = os.getenv("DASHSCOPE_API_KEY")
DASHSCOPE_BASE_URL   = os.getenv("DASHSCOPE_BASE_URL")


model = init_chat_model(
    model="openai:qwen-vl-plus",
    api_key = DASHSCOPE_API_KEY,
    base_url = DASHSCOPE_BASE_URL,
)

template = PromptTemplate.from_template("你是一位{name},请用简单的明了的语言讲解一下{js}")
prompt = template.format(name="CC",js="java")

chat_template = ChatPromptTemplate([
    ("system","你是一位python技术专家"),
    ("human","{user_input}")
    ]
)
pro = chat_template.invoke({"user_input":"讲一下python的函数"})

response = model.invoke(pro)

print(response.content)
```

2. ChatPromptTemplate模版使用（支持角色定义、system、user/human、assistant/ai）

2.1 from_message方式（推荐）

```python
template = ChatPromptTemplate.from_messages(
    [
        ("system","我是一位AI开发工程师，你的名字叫{name}"),
        ("human","{user_input}"),
    ]
)
prompt = template.invoke({"name":"CC","user_input":"你是谁"})
print(prompt)
```

2.2 实例初始化

```python
template = ChatPromptTemplate(
    [
        ("system","我是一位AI开发工程师，你的名字叫{name}"),
        ("human","{user_input}"),
    ]
)
prompt = template.invoke({"name":"CC","user_input":"你是谁"})
print(prompt)
```

2.3 模版调用3中方式

invoke方法（返回ChatPromptValue）

```python
template = ChatPromptTemplate(
    [
        ("system","我是一位AI开发工程师，你的名字叫{name}"),
        ("human","{user_input}"),
    ]
)
# 返回ChatPromptValue
prompt = template.invoke({"name":"CC","user_input":"你是谁"})
print(type(prompt))
```

format方式（返回str）

```python
format = template.format(name="CC",user_input="你是谁")
print(type(format))
```

format_message方式（返回list）

```python
mess = template.format_messages(name="CC",user_input="你是谁")
print(type(mess))
```

2.4 大模型调用，模版返回值均可作为参数给大模型

```python
res = model.invoke(mess)
print(res)
```
