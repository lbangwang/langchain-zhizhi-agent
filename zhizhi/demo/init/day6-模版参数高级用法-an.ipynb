1.部分变量预填充 partial()函数

```python


from langchain.chat_models import init_chat_model
from dotenv import load_dotenv
import os

from langchain_core.prompts import ChatPromptTemplate

load_dotenv(override=True)

DASHSCOPE_API_KEY = os.getenv("DASHSCOPE_API_KEY")
DASHSCOPE_BASE_URL   = os.getenv("DASHSCOPE_BASE_URL")


model = init_chat_model(
    model="openai:qwen-vl-plus",
    api_key = DASHSCOPE_API_KEY,
    base_url = DASHSCOPE_BASE_URL,
)

str_message = ChatPromptTemplate.from_messages([
    ("system","你是{name},服务于{aud}"),
    ("human","{user_input}"),
])

# 原始方法填充
# message = str_message.invoke({"name":"CC","aud":"程序员","user_input":"你是干什么的"})

# 使用partial函数
temp = str_message.partial(name = "CC",aud = "程序员")
messages = temp.invoke({"user_input":"你是干什么的"})

print(messages)
```

1.1 不同场景创建不同角色

```python


from langchain.chat_models import init_chat_model
from dotenv import load_dotenv
import os

from langchain_core.messages import HumanMessage
from langchain_core.prompts import ChatPromptTemplate

load_dotenv(override=True)

DASHSCOPE_API_KEY = os.getenv("DASHSCOPE_API_KEY")
DASHSCOPE_BASE_URL   = os.getenv("DASHSCOPE_BASE_URL")


model = init_chat_model(
    model="openai:qwen-vl-plus",
    api_key = DASHSCOPE_API_KEY,
    base_url = DASHSCOPE_BASE_URL,
)

str_message = ChatPromptTemplate.from_messages([
    ("system","你是{name},服务于{aud}"),
    ("human","{user_input}"),
])

# 使用partial函数
temp = str_message.partial(name = "IT技术",aud = "程序员")
messages = temp.invoke({"user_input":"你是干什么的"})
print(messages)

temp1 = str_message.partial(name = "销售支持",aud = "客户")
messages1 = temp.invoke({"user_input":"你是干什么的"})

print(messages1)
```

1.2 消息占位符 （JSON形式、messageplaceholder)

```python
json_str = ChatPromptTemplate.from_messages(
    [
        ("system","你是一个AI助手"),
        ("placeholder","{conver}"),
    ]
)

prompt = json_str.invoke({"conver":[
    ("human","你是谁"),
    ("ai","猜猜看"),
]})
print(prompt)
```

1.3 message形式

```python
from langchain_core.prompts import  MessagesPlaceholder
from langchain_core.messages.ai import AIMessage
json_str = ChatPromptTemplate.from_messages(
    [
        ("system","你是一个AI助手"),
        MessagesPlaceholder("msg"),
    ]
)

# prompt = json_str.invoke({"msg":[
#     ("human","你是谁"),
#     ("ai","猜猜看"),
# ]})

prompt = json_str.invoke({"msg":[
   HumanMessage(content="你是十恶"),
    AIMessage(content="ddd")
]})
print(prompt)
```
