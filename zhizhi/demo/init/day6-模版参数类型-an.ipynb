1.ChatPromptTemplate参数类型（str、元组、字典、消息、MessagePromptTemplate、BaseChatPromptTemplate）

1.1 str类型

2.1 from_message方式（推荐）

```python
import base64
from langchain.chat_models import init_chat_model
from dotenv import load_dotenv
import os

from langchain_core.messages import SystemMessage, HumanMessage
from langchain_core.prompts import ChatPromptTemplate, HumanMessagePromptTemplate, SystemMessagePromptTemplate

load_dotenv(override=True)

DASHSCOPE_API_KEY = os.getenv("DASHSCOPE_API_KEY")
DASHSCOPE_BASE_URL   = os.getenv("DASHSCOPE_BASE_URL")


model = init_chat_model(
    model="openai:qwen-vl-plus",
    api_key = DASHSCOPE_API_KEY,
    base_url = DASHSCOPE_BASE_URL,
)

str_message = ChatPromptTemplate.from_messages([
    "你好，我是{name}"
])

message = str_message.invoke({"name":"CC"})

print(message)
```

1.2 tuple元组类型

```python
template = ChatPromptTemplate(
    [
        ("system","我是一位AI开发工程师，你的名字叫{name}"),
        ("human","{user_input}"),
    ]
)
prompt = template.invoke({"name":"CC","user_input":"你是谁"})
print(prompt)
```

1.3 dict列表、字典类型

```python
template = ChatPromptTemplate(
    [
        {"role":"system","content":"你是{name}"},
        {"role":"user","content":"hello 你好"},
    ]
)

prompt = template.invoke({"name":"CC"})
print(prompt)
```

1.4 message类型,不能传递变量

```python
from langchain_core.messages import SystemMessage,HumanMessage
template = ChatPromptTemplate.from_messages(
    [
      SystemMessage(content="你是AI助手{name}"),
      HumanMessage(content="你哈"),
    ]
)

prompt = template.invoke({"name":"CC"})
print(prompt)
```

1.5 messagePromptTemplate列表

```python
from langchain_core.prompts import HumanMessagePromptTemplate,SystemMessagePromptTemplate

hum = HumanMessagePromptTemplate.from_template("你是谁")
sys = SystemMessagePromptTemplate.from_template("你是一个{name}")
template = ChatPromptTemplate.from_messages(
    [
      sys,
      hum,
    ]
)

prompt = template.invoke({"name":"CC"})
print(prompt)
```

1.6 BaseChatPormptTemplate类型，相当于自己传递自己

```python


hum = template = ChatPromptTemplate.from_messages(
    [
        ("system","我是一个{name}"),
    ]
)
hum = template = ChatPromptTemplate.from_messages(
    [
        ("user","你是谁"),
    ]
)
template = ChatPromptTemplate.from_messages(
    [
      sys,
      hum,
    ]
)

prompt = template.invoke({"name":"CC"})
print(prompt)
```
