## 1. Agent绑定工具

```python

from langchain.chat_models import init_chat_model
from langchain.agents import create_agent
import os

from dotenv import load_dotenv
from langchain_core.messages import HumanMessage, SystemMessage
from langchain_core.tools import tool
from langchain_tavily import TavilySearch
from openai import max_retries
from rich import print as rprint

# 加载配置文件，存在相同key采用当前覆盖
load_dotenv(override=True)

# 具体模型的key和url
DEEPSEEK_API_KEY = os.getenv("DEEPSEEK_API_KEY")
DEEPSEEK_API_BASE   = os.getenv("DEEPSEEK_BASE_URL")
DEEPSEEK_MODEL_NAME   = os.getenv("DEEPSEEK_MODEL")


@tool(parse_docstring=True)
def get_weather(city:str)->str:
    """查询天气工具

    Args:
        city : 具体城市名称
        """
    return f"{city}天气晴朗！！！30度"



model = init_chat_model(
    model=DEEPSEEK_MODEL_NAME,
    model_provider="deepseek",
    api_key = DEEPSEEK_API_KEY,
    base_url = DEEPSEEK_API_BASE,
)
# 创建agent
agent = create_agent(
    model=model,
    tools=[get_weather]
)


messages = {
    "messages":[
        HumanMessage("今天杭州天气怎么样？")
    ]
}

res = agent.invoke(messages)
rprint(res)
```

## 2.0 接入内置工具TavilySearch

```python

from langchain.chat_models import init_chat_model
from langchain.agents import create_agent
import os

from dotenv import load_dotenv
from langchain_core.messages import HumanMessage
from langchain_tavily.tavily_search import TavilySearch
from rich import print as rprint

# 加载配置文件，存在相同key采用当前覆盖
load_dotenv(override=True)

# 具体模型的key和url
DEEPSEEK_API_KEY = os.getenv("DEEPSEEK_API_KEY")
DEEPSEEK_API_BASE   = os.getenv("DEEPSEEK_BASE_URL")
DEEPSEEK_MODEL_NAME   = os.getenv("DEEPSEEK_MODEL")

web_seacher = TavilySearch(
    max_retries = 2,
    tavily_api_key= os.getenv("TAILY_SEARCH_API_KEY"),
)


model = init_chat_model(
    model=DEEPSEEK_MODEL_NAME,
    model_provider="deepseek",
    api_key = DEEPSEEK_API_KEY,
    base_url = DEEPSEEK_API_BASE,
)
# 创建agent
agent = create_agent(
    model=model,
    tools=[web_seacher]
)


messages = {
    "messages":[
        HumanMessage("请问一下2025年NBA总决赛有哪些队伍？")
    ]
}

res = agent.invoke(messages)
rprint(res)
```
