1.2 Annotated使用

```python
import os

from dotenv import load_dotenv
from langchain.chat_models.base import init_chat_model
from typing_extensions import TypedDict, Annotated


class MovieDict(TypedDict):
    title: Annotated[str,"电影名称"]
    year: Annotated[int,"发布年份"]
    director: Annotated[str,"电影导演名字"]
    rating: Annotated[float,"电影评分"]

# 加载配置文件，存在相同key采用当前覆盖
load_dotenv(override=True)

DASHSCOPE_API_KEY = os.getenv("DASHSCOPE_API_KEY")
DASHSCOPE_BASE_URL   = os.getenv("DASHSCOPE_BASE_URL")


model = init_chat_model(
    model="openai:qwen-plus",
    api_key = DASHSCOPE_API_KEY,
    base_url = DASHSCOPE_BASE_URL,
)

movie_model = model.with_structured_output(MovieDict)
res = movie_model.invoke("给我介绍一下电影星际穿越")
print(res)
```

2.0 嵌套结构

```python
import os

from dotenv import load_dotenv
from langchain.chat_models.base import init_chat_model
from typing_extensions import TypedDict, Annotated,List


class Actor(TypedDict):
    """演员信息"""
    name: Annotated[str,"演员名称"]
    role: Annotated[str,"演员角色"]


class MovieDict(TypedDict):
    title: Annotated[str,"电影名称"]
    year: Annotated[int,"发布年份"]
    director: Annotated[str,"电影导演名字"]
    rating: Annotated[float,"电影评分"]
    cast: Annotated[List[Actor], "演员列表"]

# 加载配置文件，存在相同key采用当前覆盖
load_dotenv(override=True)

DASHSCOPE_API_KEY = os.getenv("DASHSCOPE_API_KEY")
DASHSCOPE_BASE_URL   = os.getenv("DASHSCOPE_BASE_URL")


model = init_chat_model(
    model="openai:qwen-plus",
    api_key = DASHSCOPE_API_KEY,
    base_url = DASHSCOPE_BASE_URL,
)

movie_model = model.with_structured_output(MovieDict)
res = movie_model.invoke("给我介绍一下盗梦空间")
print(res)
```

1.3 ...的使用，是python的自变量，等价于Ellipsis，，这块代表此字段为必填字

```python
import os

from dotenv import load_dotenv
from langchain.chat_models.base import init_chat_model
from typing_extensions import TypedDict, Annotated,List


class MovieDict(TypedDict):
    title: Annotated[str,...,"电影名称"]
    year: Annotated[int,...,"发布年份"]
    director: Annotated[str,...,"电影导演名字"]
    rating: Annotated[float,...,"电影评分"]

# 加载配置文件，存在相同key采用当前覆盖
load_dotenv(override=True)

DASHSCOPE_API_KEY = os.getenv("DASHSCOPE_API_KEY")
DASHSCOPE_BASE_URL   = os.getenv("DASHSCOPE_BASE_URL")


model = init_chat_model(
    model="openai:qwen-plus",
    api_key = DASHSCOPE_API_KEY,
    base_url = DASHSCOPE_BASE_URL,
)

movie_model = model.with_structured_output(MovieDict)
res = movie_model.invoke("根据这段话抽取盗梦空间的信息，不包含的信息必须留空：盗梦空间上映于2010年，我们并不知道它的导演是谁。")
print(res)
```
